# 5OD1 three-histidine zinc-site search

This notebook uses the same entry point as the command-line example. Install METALPROT and select the installed environment as the Jupyter kernel. Start from the repository root or this example directory. The database must be prepared separately. This example is not the 2CBA reconstruction in the paper.


```python
from pathlib import Path
import os
import subprocess
import sys

root = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / "scripts/run_search.py").is_file()
             and (p / "examples/basic_search/config.json").is_file()), None)
if root is None:
    raise FileNotFoundError("Start Jupyter inside the METALPROT repository")
config = root / "examples/basic_search/config.json"
subprocess.run([sys.executable, str(root / "scripts/run_search.py"),
                "--config", str(config), "--check-only"], check=True)
```

```python
env = dict(os.environ, OPENBLAS_NUM_THREADS="1", OMP_NUM_THREADS="1", NUMBA_NUM_THREADS="1")
subprocess.run(["bash", str(root / "examples/basic_search/run.sh"),
                str(root / "outputs/notebook")],
               check=True, env=dict(env, PATH=str(Path(sys.executable).parent) + os.pathsep + env.get("PATH", "")))
```

The runner prints the result directory and compares the current results against the bundled reference. A successful comparison reports `design_count: 4` and `errors: []`. See [Usage](../../docs/usage.md) for output fields.
